```python
import numpy as np
from qiskit import QuantumCircuit
from qiskit.primitives import StatevectorSampler

sampler = StatevectorSampler(seed=42)

def run(qc, shots=1000):
    qc = qc.copy()
    qc.measure_all()
    return sampler.run([qc], shots=shots).result()[0].data.meas.get_counts()
```

```python
for angle in [0, np.pi/4, np.pi/2, 3*np.pi/4, np.pi]:
    qc = QuantumCircuit(1)
    qc.ry(angle, 0)
    counts = run(qc)
    ones = counts.get("1", 0) / 1000
    print(f"angle = {angle:.2f}   fraction of 1s = {ones:.3f}")
```

```python
def similarity(a, b):
    qc = QuantumCircuit(1)
    qc.ry(a, 0)     
    qc.ry(-b, 0)     
    counts = run(qc)
    return counts.get("0", 0) / 1000   

print("A vs A:", similarity(0.5, 0.5))
print("A vs B:", similarity(0.5, 0.6))
print("A vs C:", similarity(0.5, 2.8))
```
